# Task 1: Exploring and Visualizing the Iris Dataset

**Objective:** Load the Iris dataset, explore its structure, and create visualizations to see whether the 3 species can be distinguished by their measurements.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Importing libraries")
```

```python
# Loading Iris dataset from seaborn
df = sns.load_dataset('iris')

# Print first 5 rows
print('Printing first 5 rows: ')
df.head()
```

```python
# How many rows and columns?
print('Shape',df.shape)

# Columns names
print('Columns: ', df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
df['species'].value_counts()
```

```python
# Scatter Plot: Sepat Length vs Sepal Width

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='sepal_width',
    hue='species',
    palette='Set1'
)
plt.title('Sepal Length vs Sepal Width by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.legend(title='Species')
plt.tight_layout()
plt.show()
```

```python
# Histogram for all 4 measurements

df.drop('species',axis=1).hist(
    bins =20,
    figsize = (10, 6),
    color='steelblue',
    edgecolor='black',
)
plt.suptitle('Distribution of all Features', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Box Plots: Grouped by Species

fig, axes = plt.subplots(2,2,figsize=(12,8))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for i, feature in enumerate(features):
    ax = axes[i//2, i%2]
    sns.boxplot(data=df, x='species', y=feature, hue='species', palette='pastel' ,ax=ax)
    ax.set_title(f'Box Plot of {feature}')

plt.tight_layout()
plt.show()
```

# Summary
1. Dataset -> 150 Rows(flowers) and 5 Columns(pieces of information)
2. 3 Species -> Setosa, Versicolor and Virginica
3. Scatter plot -> Setosa is separated from other 2 species
4. Histogram -> Setosa has smaller petals as compared to others
5. Box plot -> Setosa are completely separate in petals and sepal measurements
